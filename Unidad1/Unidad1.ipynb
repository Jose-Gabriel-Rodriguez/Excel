```python
# Condiciones 1
variable1 = 30
variable2 = 50
if variable1 > variable2:
    print("La variable 1 es mayor que la variable 2")
else:
    print("La variable 2 es mayor que la variable 1")
```

```python
# Condiciones 2
variable1 = 70
variable2 = 50
if variable1 > variable2:
    print( variable1, "es mayor que", variable2 )
else:
    print(variable2, "es mayor que", variable1)
```

```python
#  Condiciones 3    usando elif  dentro de if
variable1 = 70
variable2 = 70
if variable1 > variable2:
    print( variable1, "es mayor que", variable2 )
elif variable1 == variable2:
    print(variable1, "es igual que", variable2)
else:
    print(variable2, "es mayor que", variable1)
print("Fin")
```

```python
# ******************************* Tema de ciclos   FOR
# Programita 1 de tablas de multiplicar
tabla = 5
for i in range(10):   # range con 1 parametro es de 0 a n-1
    print(tabla, "x", i, "=", tabla*i)
```

```python
# Programita 2 de tablas de multiplicar
tabla = 5
for i in range(1,11):   # range con 2 parametros
    print(tabla, "x", i, "=", tabla*i)
```

```python
# Programita 3 de tablas de multiplicar
tabla = 5
for i in range(10,31,2):   # range con 3 parametros (incrementos )
    print(tabla, "x", i, "=", tabla*i)
```

```python
####
#### ********************** RETO *******************
# Hacer un programa que me pida la tabla de multiplicar y el limite inicial y final

tabla = int(input("Dame la tabla de multiplicar "))
limite_inicial = int(input("Dame el limite inicial "))
limite_final = int(input("Dame el limite final "))

for i in range(limite_inicial, limite_final+1 ):
    print(tabla, "x", i, "=", tabla*i)
```

```python
###############   demostracion de while

minimo = 1
maximo = 10
numero = int(input(f"Ingrese un número entre {minimo} y {maximo}: "))
while numero < minimo or numero > maximo:
    print("Número fuera de rango. Intente de nuevo.")
    numero = int(input(f"Ingrese un número entre {minimo} y {maximo}: "))

print(f"Número válido: {numero}")
```

```python
listaAlumnos = ["Ana", "Luis", "Carlos"]
print("Alumnos =", listaAlumnos)
```

```python
# Accedemos a los valores por medio de indices

print("Alumno en la posicion 0 =", listaAlumnos[0])
print("Alumno en la posicion 1 =", listaAlumnos[1])
print("Alumno en la posicion 2 =", listaAlumnos[2])
```

```python
# Modificar un Valor de la listaAlumnos
# Lo hacemos por medio del indice
print("Alumnos =", listaAlumnos)
listaAlumnos[1] = "Juan"
print("Alumnos =", listaAlumnos)
```

```python
# Agregar un valor a la lista con Append
# la funcion append lo agrega al final de la lista
listaAlumnos.append("Maria")
print("Alumnos =", listaAlumnos)
```

```python
# Agregar un valor a la lista con la funcion insert
# insert lo agrega en la posicion especificada y recorre los demás
listaAlumnos.insert(2, "Pedro")
print("Alumnos =", listaAlumnos)
```

```python
# Borrrar un item o lelemnto de la lista con REMOVE
# Marcara error porque juan no existe,
# en la lista, esta la inicial con mayuscula
listaAlumnos.remove("juan")
```

```python
# Ahora si lo borramos tecleando bien su nombre
listaAlumnos.remove("Juan")
print("Alumnos =", listaAlumnos)
```

```python
# Borrra por medio de su posicion en la lista con POP
# en este caso se borra Pedro que es el que esta en la posicion 1
listaAlumnos.pop(1)
print("Alumnos =", listaAlumnos)
```

```python
# Borrra por medio de su posicion en la lista con DEL
# en este caso se borra Pedro que es el que esta en la
# posicion 1 en la nueva lista
del listaAlumnos[1]
print("Alumnos =", listaAlumnos)
```

```python
#    FUNCION LEN
print("Numero de Alumnos en la lista", len(listaAlumnos))
```

```python
listaAlumnos.append("Pedro")
listaAlumnos.append("Rebeca")
print("Alumnos =", listaAlumnos)
```

```python
# Recorrer una lista de valores con la instruccion FOR
# La variable alumno toma el valor de cada elemento, uno por uno.
# No es necesario manejar directamente los índices
for alumno in listaAlumnos:
  print(alumno)
```

```python
# Recorrer una lista por medio de sus indices
for i in range( len(listaAlumnos) ):
  print(listaAlumnos[i])
```

```python
# Buscar en una lista
# En este caso esta mal capturado el valor (minusculas)
software = ["Python", "Java", "C++", "AutoCAD", "SolidWorks"]
if  "python" in software:
  print("Diponible")
else:
  print("No disponible")
```

```python
# Posible solucion
# generar nueva lista y comparar en minusculas

software = ["Python", "Java", "C++", "AutoCAD", "SolidWorks"]
buscarPrograma="pyTHon"

# Convierte todos los nombres de software a minúsculas para
# realizar una búsqueda sin distinción de mayúsculas y minúsculas.
software_minusculas = [item.lower() for item in software]

# Ahora, comparamos la variable de búsqueda en minúsculas
# con la lista de software en minúsculas.
if buscarPrograma.lower() in software_minusculas:
  print("Disponible")
else:
  print("No disponible")

print("Fin")
```
