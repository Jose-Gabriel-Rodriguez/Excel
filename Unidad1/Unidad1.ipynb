```python
# Condiciones 1
variable1 = 30
variable2 = 50
if variable1 > variable2:
    print("La variable 1 es mayor que la variable 2")
else:
    print("La variable 2 es mayor que la variable 1")
```

```python
# Condiciones 2
variable1 = 70
variable2 = 50
if variable1 > variable2:
    print( variable1, "es mayor que", variable2 )
else:
    print(variable2, "es mayor que", variable1)
```

```python
#  Condiciones 3    usando elif  dentro de if
variable1 = 70
variable2 = 70
if variable1 > variable2:
    print( variable1, "es mayor que", variable2 )
elif variable1 == variable2:
    print(variable1, "es igual que", variable2)
else:
    print(variable2, "es mayor que", variable1)
print("Fin")
```

```python
# ******************************* Tema de ciclos   FOR
# Programita 1 de tablas de multiplicar
tabla = 5
for i in range(10):   # range con 1 parametro es de 0 a n-1
    print(tabla, "x", i, "=", tabla*i)
```

```python
# Programita 2 de tablas de multiplicar
tabla = 5
for i in range(1,11):   # range con 2 parametros
    print(tabla, "x", i, "=", tabla*i)
```

```python
# Programita 3 de tablas de multiplicar
tabla = 5
for i in range(10,31,2):   # range con 3 parametros (incrementos )
    print(tabla, "x", i, "=", tabla*i)
```

```python
####
#### ********************** RETO *******************
# Hacer un programa que me pida la tabla de multiplicar y el limite inicial y final

tabla = int(input("Dame la tabla de multiplicar "))
limite_inicial = int(input("Dame el limite inicial "))
limite_final = int(input("Dame el limite final "))

for i in range(limite_inicial, limite_final+1 ):
    print(tabla, "x", i, "=", tabla*i)
```
