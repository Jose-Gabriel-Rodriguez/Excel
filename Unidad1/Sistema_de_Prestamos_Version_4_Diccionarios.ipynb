```python
# SISTEMA DE PRÉSTAMOS
# Versión 4 - Uso de listas y diccionarios

import datetime

# Lista de alumnos: cada alumno es un diccionario
alumnos = [
    {"numero_control": "26040200", "nombre": "Superman","carrera": "Ingeniería en Inteligencia Artificial"},
    {"numero_control": "26040201", "nombre": "Bat Man", "carrera": "Ingeniería en Inteligencia Artificial"},
    {"numero_control": "26040202", "nombre": "Iron Man", "carrera": "Ingeniería en Inteligencia Artificial"},
    {"numero_control": "26040203", "nombre": "Linterna Verde", "carrera": "Ingeniería Informática"},
    {"numero_control": "26040204", "nombre": "Hulk", "carrera": "Ingeniería Informática"}
]

# Lista de computadoras: cada computadora es un diccionario
computadoras = [
    {"id": "PC-LC1-01", "aula": "LC1", "software": "Python, VS Code", "estado": "Disponible"},
    {"id": "PC-LC1-02", "aula": "LC1", "software": "Python, Java, VS Code", "estado": "Prestada"},
    {"id": "PC-LC1-03", "aula": "LC1", "software": "AutoCAD, MATLAB", "estado": "Reservada"},
    {"id": "PC-LC2-01", "aula": "LC2", "software": "Python, VS Code", "estado": "Disponible"},
    {"id": "PC-LC2-02", "aula": "LC2", "software": "Python, Java, VS Code", "estado": "Prestada"},
    {"id": "PC-LC2-03", "aula": "LC2", "software": "AutoCAD, MATLAB", "estado": "Disponible"},
    {"id": "PC-LC3-01", "aula": "LC3", "software": "Python, VS Code", "estado": "Disponible"},
    {"id": "PC-LC3-02", "aula": "LC3", "software": "Python, Java, VS Code", "estado": "Prestada"},
    {"id": "PC-LC3-03", "aula": "LC3", "software": "AutoCAD, MATLAB", "estado": "Reservada"},
    {"id": "PC-LC3-04", "aula": "LC3", "software": "Python, VS Code", "estado": "Mantenimiento"}
]

# Historial inicial de préstamos. Cada préstamo es un diccionario.
prestamos = [
    {"numero_control": "26040201", "software": "python", "aula": "LC1",
     "computadora": "PC-LC1-01", "fecha": "Fecha no registrada"},
    {"numero_control": "26040202", "software": "java", "aula": "LC1",
     "computadora": "PC-LC1-02", "fecha": "Fecha no registrada"},
    {"numero_control": "26040201", "software": "python", "aula": "LC2",
     "computadora": "PC-LC2-01", "fecha": "Fecha no registrada"},
    {"numero_control": "26040203", "software": "matlab", "aula": "LC3",
     "computadora": "PC-LC3-01", "fecha": "Fecha no registrada"},
    {"numero_control": "26040204", "software": "python", "aula": "LC3",
     "computadora": "PC-LC3-02", "fecha": "Fecha no registrada"}
]

# Lista de necesidades de software
necesidades = [
    {"software": "Photoshop", "fecha": "12/10/2026 07:30:00"}
]


def consultar_computadoras():
    print("\n--- COMPUTADORAS ---")
    for computadora in computadoras:
        print("Aula:", computadora["aula"],
              "| ID:", computadora["id"],
              "| Software:", computadora["software"],
              "| Estado:", computadora["estado"])


def buscar_sw():
    programa = input("\nIngrese el software que desea buscar: ").strip()
    print("\n--- RESULTADOS ---")
    encontrado = False

    for computadora in computadoras:
        if programa.lower() in computadora["software"].lower():
            print("ID:", computadora["id"],
                  "| Aula:", computadora["aula"],
                  "| Software:", computadora["software"],
                  "| Estado:", computadora["estado"])
            encontrado = True

    if not encontrado:
        print("No se encontraron computadoras con ese software.")


def solicitar_prestamo():
    print("\n=== SOLICITAR PRÉSTAMO ===")
    num_control = input("Número de control del alumno: ").strip()

    alumno_encontrado = False
    for alumno in alumnos:
        if num_control == alumno["numero_control"]:
            alumno_encontrado = True
            break

    if not alumno_encontrado:
        print("El alumno no existe en la lista.")
        return

    programa = input("¿Qué programa necesita?: ").strip()
    software_encontrado = False
    computadora_asignada = False

    for computadora in computadoras:
        if programa.lower() in computadora["software"].lower():
            software_encontrado = True

            if computadora["estado"] == "Disponible":
                computadora["estado"] = "Prestada"
                fecha_actual = datetime.datetime.now().strftime("%d/%m/%Y %H:%M:%S")

                prestamo = {
                    "numero_control": num_control,
                    "software": programa,
                    "aula": computadora["aula"],
                    "computadora": computadora["id"],
                    "fecha": fecha_actual
                }
                prestamos.append(prestamo)

                print("Se asignó la computadora", computadora["id"],
                      "que está en el aula", computadora["aula"])
                computadora_asignada = True
                break
            else:
                print("Se encontró el software, pero una computadora está:",
                      computadora["estado"])

    if not software_encontrado:
        print("No se encontró ese software.")
        opcion = input("¿Deseas agregarlo a la lista de necesidades? (s/n): ")
        if opcion.lower() == "s":
            fecha_actual = datetime.datetime.now().strftime("%d/%m/%Y %H:%M:%S")
            necesidades.append({"software": programa, "fecha": fecha_actual})
            print("Necesidad de software registrada.")
    elif not computadora_asignada:
        print("No se encontró una computadora disponible con ese software.")


def devolver_computadora():
    print("\n=== DEVOLVER COMPUTADORA ===")
    id_pc = input("Dame el ID de la computadora: ").strip()
    encontrada = False

    for computadora in computadoras:
        if computadora["id"] == id_pc:
            encontrada = True

            if computadora["estado"] == "Prestada" or computadora["estado"] == "Reservada":
                computadora["estado"] = "Disponible"
                print("La computadora", id_pc, "ya está disponible.")
            else:
                print("La computadora", id_pc, "no está prestada ni reservada.")
            break

    if not encontrada:
        print("No existe la computadora", id_pc)


def crear_reservacion():
    print("\n--- CREAR RESERVACIÓN ---")
    print("Esta opción se implementará posteriormente.")


def agregar_computadora():
    print("\n--- AGREGAR COMPUTADORA ---")
    id_computadora = input("ID de la computadora: ").strip()

    # Evitar registrar dos computadoras con el mismo ID
    for computadora in computadoras:
        if computadora["id"].lower() == id_computadora.lower():
            print("Ya existe una computadora con ese ID.")
            return

    aula = input("Aula: ").strip()
    programas = input("Software instalado: ").strip()

    nueva_pc = {
        "id": id_computadora,
        "aula": aula,
        "software": programas,
        "estado": "Disponible"
    }
    computadoras.append(nueva_pc)
    print("\nComputadora agregada correctamente.")


def historial_prestamos():
    print("\n--- HISTORIAL DE PRÉSTAMOS DE COMPUTADORAS ---")

    if len(prestamos) > 0:
        for i, prestamo in enumerate(prestamos, start=1):
            print(i,
                  "| Alumno:", prestamo["numero_control"],
                  "| Software:", prestamo["software"],
                  "| Aula:", prestamo["aula"],
                  "| ID:", prestamo["computadora"],
                  "| Fecha:", prestamo["fecha"])
    else:
        print("No se han registrado préstamos.")


def estadisticas():
    # Diccionario para contar préstamos por carrera
    conteo_carreras = {}

    for prestamo in prestamos:
        carrera = "Carrera no encontrada"

        for alumno in alumnos:
            if alumno["numero_control"] == prestamo["numero_control"]:
                carrera = alumno["carrera"]
                break

        if carrera in conteo_carreras:
            conteo_carreras[carrera] += 1
        else:
            conteo_carreras[carrera] = 1

    print("\n================================")
    print("Estadísticas por carrera:")
    if conteo_carreras:
        for carrera, cantidad in conteo_carreras.items():
            print(carrera, ":", cantidad, "veces")
    else:
        print("No hay préstamos registrados.")

    # Diccionario para contar préstamos por software solicitado
    conteo_programas = {}

    for prestamo in prestamos:
        programa = prestamo["software"].lower()

        if programa in conteo_programas:
            conteo_programas[programa] += 1
        else:
            conteo_programas[programa] = 1

    print("\nEstadísticas por programa:")
    if conteo_programas:
        for programa, cantidad in conteo_programas.items():
            print(programa, ":", cantidad, "veces")
    else:
        print("No hay préstamos registrados.")


def historico_necesidades():
    print("\n--- HISTORIAL DE NECESIDADES DE SOFTWARE ---")

    if necesidades:
        for necesidad in necesidades:
            print("Programa:", necesidad["software"],
                  "| Fecha:", necesidad["fecha"])
    else:
        print("No se han registrado necesidades de software.")


# MENÚ PRINCIPAL
opcion = 0
while opcion != 10:
    print("\n================================")
    print("      SISTEMA DE PRÉSTAMOS")
    print("================================")
    print("1. Consultar computadoras")
    print("2. Buscar por software")
    print("3. Solicitar préstamo")
    print("4. Devolver computadora")
    print("5. Crear reservación")
    print("6. Agregar computadora")
    print("7. Historial de préstamos")
    print("8. Necesidades de software")
    print("9. Estadísticas")
    print("10. Salir")
    print("================================")

    try:
        opcion = int(input("Seleccione una opción: "))
    except ValueError:
        print("Debe ingresar un número.")
        continue

    if opcion == 1:
        consultar_computadoras()
    elif opcion == 2:
        buscar_sw()
    elif opcion == 3:
        solicitar_prestamo()
    elif opcion == 4:
        devolver_computadora()
    elif opcion == 5:
        crear_reservacion()
    elif opcion == 6:
        agregar_computadora()
    elif opcion == 7:
        historial_prestamos()
    elif opcion == 8:
        historico_necesidades()
    elif opcion == 9:
        estadisticas()
    elif opcion == 10:
        print("\nGracias por utilizar el sistema.")
    else:
        print("\nOpción no válida.")
```
