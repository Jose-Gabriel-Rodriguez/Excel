```python
# SISTEMA DE PRÉSTAMOS
# Versión 4 - Listas y diccionarios

# Lista de alumnos activos en el sistema
alumnos = [ "26040200", "26040201", "26040202", "26040203", "26040204"]

# LISTA DE COMPUTADORAS
# Cada computadora es un diccionario
computadoras = [
      {"id": "PC-LC1-01", "aula": "LC1", "software": "Python, VS Code", "estado": "Disponible"},
      {"id": "PC-LC1-02", "aula": "LC1", "software": "Python, Java, VS Code", "estado": "Prestada"},
      {"id": "PC-LC1-03", "aula": "LC1", "software": "AutoCAD, MATLAB", "estado": "Reservada"},
      {"id": "PC-LC2-01", "aula": "LC2", "software": "Python, VS Code", "estado": "Disponible"},
      {"id": "PC-LC2-02", "aula": "LC2", "software": "Python, Java, VS Code", "estado": "Prestada"},
      {"id": "PC-LC2-03", "aula": "LC2", "software": "AutoCAD, MATLAB", "estado": "Reservada"},
      {"id": "PC-LC3-01", "aula": "LC3", "software": "Python, VS Code", "estado": "Disponible"},
      {"id": "PC-LC3-02", "aula": "LC3", "software": "Python, Java, VS Code", "estado": "Prestada"},
      {"id": "PC-LC3-03", "aula": "LC3", "software": "AutoCAD, MATLAB", "estado": "Reservada"},
      {"id": "PC-LC3-04", "aula": "LC3", "software": "Python, VS Code", "estado": "Mantenimiento"}
]

# CONSULTAR COMPUTADORAS
def consultar_computadoras():
    print("\n--- COMPUTADORAS ---")
    for computadora in computadoras:
        print("Aula:", computadora["aula"], "| ID:", computadora["id"], "| Software:",
              computadora["software"], "| Estado:", computadora["estado"]  )



# BUSCAR POR SOFTWARE
def buscar_por_software():
    programa = input("\nIngrese el software que desea buscar: ")
    print("\n--- RESULTADOS ---")
    encontrado = False
    for computadora in computadoras:
        if programa.lower() in computadora["software"].lower():
            print("ID:", computadora["id"], "1 Aula:", computadora["aula"],
                  "| Software:", computadora["software"], "| Estado:", computadora["estado"])
            encontrado = True
    if encontrado == False:
        print("No se encontraron computadoras con ese software.")


# SOLICITAR PRÉSTAMO
def solicitar_prestamo():
    print("\n--- SOLICITAR PRÉSTAMO ---")
    num_control = input("\nIngrese el número de control: ")
    if num_control in alumnos:
        id_computadora = input("Ingrese el ID de la computadora: ")
        encontrado = False
        for computadora in computadoras:
            if computadora["id"] == id_computadora:
                encontrado = True
                if computadora["estado"] == "Disponible":
                    computadora["estado"] = "Prestada"
                    print("\nPréstamo registrado correctamente.")
                    print("Computadora:", computadora["id"])
                    print("Aula:", computadora["aula"])
                else:
                    print("\nLa computadora no está disponible.")
        if encontrado == False:
            print("\nLa computadora no existe.")
    else:
        print("\nEl alumno no existe.")


# DEVOLVER COMPUTADORA
def devolver_computadora():
    print("\n--- DEVOLVER COMPUTADORA ---")
    id_computadora = input("Ingrese el ID de la computadora: ")
    encontrado = False
    for computadora in computadoras:
        if computadora["id"] == id_computadora:
            encontrado = True
            if computadora["estado"] == "Prestada" or computadora["estado"] == "Reservada":
                computadora["estado"] = "Disponible"
                print("\nComputadora devuelta correctamente.")
            else:
                print("\nLa computadora no se encuentra prestada o reservada.")
    if encontrado == False:
        print("\nLa computadora no existe.")


# CREAR RESERVACIÓN
def crear_reservacion():
    print("\n--- CREAR RESERVACIÓN ---")
    print("Esta opción se implementará posteriormente.")


# AGREGAR COMPUTADORA
def agregar_computadora():
    print("\n--- AGREGAR COMPUTADORA ---")
    id_computadora = input("ID de la computadora: ")
    aula = input("Aula: ")
    programas = input("Software instalado: ")
    nueva_computadora = {
        "id": id_computadora,
        "aula": aula,
        "software": programas,
        "estado": "Disponible"
    }
    computadoras.append(nueva_computadora)
    print("\nComputadora agregada correctamente.")


# MENÚ PRINCIPAL
opcion = 0
while opcion <= 6:
    print("\n================================")
    print("      SISTEMA DE PRÉSTAMOS")
    print("================================")
    print("1. Consultar computadoras")
    print("2. Buscar por software")
    print("3. Solicitar préstamo")
    print("4. Devolver computadora")
    print("5. Crear reservación")
    print("6. Agregar computadora")
    print("7. Salir")
    print("================================")
    opcion = int(input("Seleccione una opción: "))
    if opcion == 1:
        consultar_computadoras()
    elif opcion == 2:
        buscar_por_software()
    elif opcion == 3:
        solicitar_prestamo()
    elif opcion == 4:
        devolver_computadora()
    elif opcion == 5:
        crear_reservacion()
    elif opcion == 6:
        agregar_computadora()
    elif opcion == 7:
        print("\nGracias por utilizar el sistema.")
    else:
        print("\nOpción no válida.")

```
